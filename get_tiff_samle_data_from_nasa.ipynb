# NASA’s Harmonized Landsat and Sentinel‑2 (HLS) dataset, available via public URL

1. Download Sample TIFF Data

```python
import requests

# Example HLS file URL (replace with your region of interest)
url = "https://data.lpdaac.earthdatacloud.nasa.gov/path/to/sample_hls.tif"
output_file = "sample_hls.tif"

# Download the file
response = requests.get(url)
with open(output_file, "wb") as f:
    f.write(response.content)

print("Satellite TIFF downloaded successfully!")
```

2. Load TIFF with Rasterio

```python
import rasterio
import numpy as np
import matplotlib.pyplot as plt

# Open the TIFF file
dataset = rasterio.open("sample_hls.tif")

# Read all bands
image = dataset.read()  # shape: (bands, height, width)
print("Bands:", dataset.count)
print("Image shape:", image.shape)

# Display RGB composite
rgb = np.stack([image[2], image[1], image[0]], axis=-1)  # R,G,B bands
plt.imshow(rgb / np.max(rgb))
plt.title("Satellite RGB Composite")
plt.show()
```
